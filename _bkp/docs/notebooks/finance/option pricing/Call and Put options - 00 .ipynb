```python
import numpy as np
from scipy.stats import norm
```

```python
def bs_call(S, K, T, r, q, sigma):
    if T <= 0: return max(0, S - K)
    d1 = (np.log(S / K) + (r - q + 0.5 * sigma**2) * T) / (sigma * np.sqrt(T))
    d2 = d1 - sigma * np.sqrt(T)
    return S * np.exp(-q * T) * norm.cdf(d1) - K * np.exp(-r * T) * norm.cdf(d2)
```

```python

# --- 1. THE HORIZON PROBLEM (Image 1) ---
# You think stock goes from 100 to 120.
S_start, S_end = 100, 120
K = 110
# You buy a 3-month option (T=0.25)
price_today = bs_call(S_start, K, 0.25, 0.05, 0.0, 0.2)

# Scenario: The move to 120 happens exactly at 3 months (Expiration)
val_at_exp = max(0, S_end - K)
print(f"Option bought for: ${price_today:.2f}")
print(f"Value if move happens AT expiration: ${val_at_exp:.2f}")
print(f"Profit/Loss: ${val_at_exp - price_today:.2f}")

# --- 2. THE DIVIDEND ANNOUNCEMENT (Image 2) ---
# Same option, but suddenly company announces a 5% dividend (q goes from 0 to 0.05)
price_with_div = bs_call(S_start, K, 0.25, 0.05, 0.05, 0.2)
print(f"\nPrice BEFORE dividend news: ${price_today:.2f}")
print(f"Price AFTER dividend news:  ${price_with_div:.2f}")
print(f"Instant loss on news: ${price_today - price_with_div:.2f}")
```
